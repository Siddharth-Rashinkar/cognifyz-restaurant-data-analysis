## Level 1 Analysis Project

```python
import pandas as pd

df = pd.read_csv(r"E:\Data_Analytics_Projects\Cognifyz-Restaurant-Data-Analysis\Dataset\Dataset.csv")

df.head()
```

```python
df.head()
```

# Level 1 — Task 1

## Top Cuisines Analysis

```python
df['Cuisines'].isnull().sum()
```

```python
cuisines = df['Cuisines'].dropna()
```

```python
all_cuisines = cuisines.str.split(', ')
```

```python
cuisine_list = all_cuisines.explode()
```

```python
top_cuisines = cuisine_list.value_counts()

top_cuisines.head(10)
```

```python
top_3 = top_cuisines.head(3)

top_3
```

```python
total_restaurants = len(df)

percentage_top3 = (top_3 / total_restaurants) * 100

percentage_top3
```

```python
percentage_top3.round(2)
```

### Data Visualization

```python
import matplotlib.pyplot as plt

top_3.plot(kind='bar')

plt.title('Top 3 Most Common Cuisines')
plt.xlabel('Cuisine')
plt.ylabel('Number of Restaurants')

plt.show()
```

Insights:

1. North Indian cuisine is the most common cuisine in the dataset.
2. Chinese cuisine is the second most preferred cuisine.
3. Fast Food restaurants are also highly popular among customers.
4. These cuisines dominate the restaurant market in the dataset.

# Level 1 — Task 2

## City Analysis

```python
city_counts = df['City'].value_counts()

city_counts.head(10)
```

```python
highest_city = city_counts.idxmax()

highest_count = city_counts.max()

print("City with highest restaurants:", highest_city)
print("Number of restaurants:", highest_count)
```

```python
avg_rating_city = df.groupby('City')['Aggregate rating'].mean()

avg_rating_city.head(10)
```

```python
highest_avg_city = avg_rating_city.idxmax()

highest_avg_rating = avg_rating_city.max()

print("City with highest average rating:", highest_avg_city)
print("Average rating:", highest_avg_rating)
```

### Data Visualization

```python
top_cities = city_counts.head(10)

top_cities.plot(kind='bar')

plt.title('Top 10 Cities by Restaurant Count')
plt.xlabel('City')
plt.ylabel('Number of Restaurants')

plt.show()
```

Insights:

1. The city with the highest number of restaurants dominates the dataset.
2. Restaurant distribution is concentrated in a few major cities.
3. Some cities have significantly higher average ratings than others.
4. Customer satisfaction differs across locations.

# Level 1 — Task 3

## Price Range Distribution

```python
price_counts = df['Price range'].value_counts()

price_counts
```

```python
price_percentage = (price_counts / len(df)) * 100

price_percentage.round(2)
```

```python
price_counts.plot(kind='bar')

plt.title('Distribution of Price Ranges')
plt.xlabel('Price Range')
plt.ylabel('Number of Restaurants')

plt.show()
```

Insights:

2. High-price restaurants are comparatively fewer.
3. Affordable dining options dominate the dataset.
4. Customer demand appears higher for budget-friendly restaurants.

## Task 4 - Restaurant Chains Analysis

```python
chain_counts = df['Restaurant Name'].value_counts()

chain_counts.head(20)
```

```python
chains = chain_counts[chain_counts > 1]

chains.head(20)
```

```python
chains.sort_values(ascending=False).head(10)
```

```python
chain_ratings = df.groupby('Restaurant Name')['Aggregate rating'].mean()

chain_ratings.loc[chains.index].sort_values(ascending=False).head(10)
```

```python
chain_votes = df.groupby('Restaurant Name')['Votes'].mean()

chain_votes.loc[chains.index].sort_values(ascending=False).head(10)
```

```python
chains.sort_values(ascending=False).head(10).plot(kind='bar')

plt.title('Top Restaurant Chains')
plt.xlabel('Restaurant Name')
plt.ylabel('Number of Outlets')

plt.show()
```
